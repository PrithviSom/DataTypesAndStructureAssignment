1.What are data structures, and why are they important?

Ans : A data structure is a specialized way of organizing and storing data in a computer so that it can be accessed and modified efficiently. It defines the relationship between data and the operations that can be performed on it.

-  Efficient Data Management: Helps organize data logically for easy access, modification, and storage.
-  Improved Performance: Enables faster execution of algorithms by choosing the right structure (e.g., hash tables for quick lookups).
-  Simplifies Complex Problems: Makes it easier to model and solve real-world problems like routing, scheduling, and searching.
-  Supports Reusability: Promotes modular and reusable code through well-defined structures.
-  Optimizes Memory Usage: Ensures data is stored compactly and accessed with minimal overhead.
-  Foundation for Algorithms: Many algorithms are built around specific data structures (e.g., trees for hierarchical data, graphs for networks).
-  Enables Scalability: Essential for handling large datasets and building scalable applications.





2.  Explain the difference between mutable and immutable data types with examples.

ans : In Python, mutable data types can be changed after creation, meaning their contents can be modified. Examples include lists, dictionaries, and sets. You can add, update, or remove elements without creating a new object.

On the other hand, immutable data types cannot be altered once created. Any change results in a new object. Examples include strings, tuples, integers, and floats. Understanding this distinction is crucial for writing efficient and bug-free code, especially when passing data between functions or using it in collections.



3. What are the main differences between lists and tuples in Python?

ans: In Python, lists and tuples are both used to store collections of items, but they differ in key ways. Lists are mutable, meaning their elements can be changed, added, or removed after creation. Tuples are immutable, so once defined, their contents cannot be altered. Lists use square brackets [ ], while tuples use parentheses ( ). Lists offer more built-in methods and are ideal for dynamic data. Tuples are faster and better suited for fixed data that shouldn't change.


4. Describe how dictionaries store data.

Keys must be unique and immutable, while values can be of any data type. This structure makes dictionaries ideal for storing structured, searchable data like user profiles or configurations.

7. How do tuples ensure data integrity in Python?

ans: Tuples ensure data integrity in Python by being **immutable**, meaning their contents cannot be changed once defined. This immutability prevents accidental modifications, making tuples ideal for storing fixed collections of values like coordinates, configuration settings, or function returns. Because they can't be altered, tuples are also **hashable**, allowing them to be used as keys in dictionaries or elements in sets. This stability makes tuples a reliable choice when consistency and protection of data are essential.


8. What is a hash table, and how does it relate to dictionaries in Python?

ans: A hash table is a data structure that stores key-value pairs and uses a hash function to compute an index (or hash code) where each key-value pair is placed. This allows for fast data retrieval, typically in constant time—O(1)—for operations like lookup, insertion, and deletion.

In Python, the dictionary is a built-in implementation of a hash table. When you create a dictionary like {'name': 'Alice'}, Python internally hashes the key 'name' to determine where to store the value 'Alice'. This is why dictionary keys must be immutable and hashable—so their hash values remain consistent.

17. Describe the time complexity of accessing elements in a dictionary?

ans: Accessing elements in a Python dictionary has an average-case **time complexity of O(1)**, thanks to its underlying **hash table** implementation. This means retrieving a value by key is typically very fast, regardless of the dictionary’s size. However, in rare cases like hash collisions or poor hash distribution, the worst-case time complexity can degrade to **O(n)**—though Python’s internal optimizations make this extremely uncommon.

Use them when you don’t need key-value mapping—just a collection of values.

19.  Why are dictionaries considered unordered, and how does that affect data retrieval?

ans: Dictionaries were traditionally considered unordered because they didn’t guarantee the order of key-value pairs during storage or iteration. This meant that the sequence in which items were added wasn’t preserved, making them unsuitable for tasks requiring strict order.


20.  Explain the difference between a list and a dictionary in terms of data retrieval.

ans : The key difference in data retrieval between a list and a dictionary lies in how you access elements:
-  Lists use index-based access, meaning you retrieve items by their position (e.g., my_list[2] gets the third item). This is great for ordered sequences but slower for searching specific values.
-  Dictionaries use key-based access, allowing you to retrieve values directly via a unique key (e.g., my_dict['name']). This is much faster and more efficient, especially for large datasets.
